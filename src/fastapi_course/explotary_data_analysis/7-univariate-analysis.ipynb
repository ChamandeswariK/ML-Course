# 1. Categorical Data

a. Countplot

```python
sns.countplot(x='Survived', data = df)
plt.title('Titanic Survived Distribution')
plt.xlabel('Survived (0=No, 1=Yes)')
plt.ylabel("Count")
plt.show()
```

```python
df['Pclass'].value_counts().plot(kind='bar')
```

b. PieChart

```python
df['Survived'].value_counts().plot(kind='pie', autopct='%.2f')
```
